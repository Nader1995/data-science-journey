## Filtering Data

Let's deal with a more interesting data set:

```python
bios = pd.read_csv('https://raw.githubusercontent.com/Nader1995/complete-pandas-tutorial/refs/heads/master/data/bios.csv')
```

Just like maps in math, there are some maps from current dataset to an ideal dataset.

- It is also possible to use .apply() method

```python
def remean_points(row):
    
    row.USDprice = row.USDprice - USDprice_mean
    return row

coffee.apply(remean_points, axis='columns')
```

* .map() and .apply() do NOT modify the original dataset

## Grouping and Sorting

To put our dataset in groups based on a specific quality, and then perform an action:

General rule:
        
        df.groupby(
        
        'column_to_group'
        )['column_to_analyze'].function()

or 

        df.groupby(
        
        'column_to_group'
        ).column_to_analyze.function()

For example:

```python
display(

    coffee
)
display(

    coffee.groupby('Type').USDprice.mean()
) # shows the mean of the 'USDprice' column for each unique value in the 'Type' column of the coffee dataframe using the groupby method

display(

    coffee.groupby('Type').USDprice.agg(['mean', 'min', 'max', 'count'])
) # shows the mean, minimum, maximum, and count of the 'USDprice' column for each unique value in the 'Type' column of the coffee dataframe using the groupby method
```

* **Multi Index** appears when we categorize a dataset based on more than one factor:

```python
display(

    coffee.groupby(['Day', 'Type']).USDprice.mean()
) # shows the mean of the 'USDprice' column for each unique combination of values in the 'Day' and 'Type' columns of the coffee dataframe using the groupby method

display(

    coffee.groupby(['Type', 'Day']).USDprice.mean()
) # shows the mean of the 'USDprice' column for each unique combination of values in the 'Type' and 'Day' columns of the coffee dataframe using the groupby method
```

* If we do NOT like **Multi Indexing**, we can always .reset_index:

```python
display(

    coffee.groupby(['Type', 'Day']).USDprice.mean().reset_index()
) # shows the mean of the 'USDprice' column for each unique combination of values in the 'Type' and 'Day' columns of the coffee dataframe using the groupby method and resets the index to a default integer index
```

* There is a problem with saving data using .to_csv() or others. It saves the index as well, so next time when we read the file, pandas adds a new index column next to the original index column the file already has, so we have to save dataset without saving the index column, using index=false parameter:

```python
coffee.to_csv('no_index_coffee.csv', index= False) # saves the coffee dataframe to a CSV file without the index
```

In sorting, we want to re-arrange rows based on some values in a column. 
It is usually in ascending order but you can modify by writing:

    ascending = False

Here is an example:

```python
display(

    coffee.sort_values(by='Units Sold')
) # sorts the coffee dataframe by the 'Units Sold' column in ascending order

display(

    coffee.sort_values(by='Units Sold', ascending= False)
) # sorts the coffee dataframe by the 'Units Sold' column in descending order
```

* Or we can choose more than one parameter to sort:

```python
display(

    coffee.sort_values(by=['Type', 'Units Sold'])
) # sorts the coffee dataframe by the 'Type' column in ascending order and then by the 'Units Sold' column in ascending order within each group of 'Type'
```

## Data Types and Missing Values

* We can use .dtype method to know the type of data of a specific column or the whole dataset:

```python
display(coffee.USDprice.dtype) # shows the data type of the 'USDprice' column in the coffee dataframe
display('********************')
display(coffee.dtypes) # shows the data types of all columns in the coffee dataframe
```

In order to change data type:

```python
display(coffee.USDprice.astype('float64')) # converts the 'USDprice' column of the coffee dataframe to the float64 data type using the astype method
```

Missing values: 

We can use 

    .isnull()

to find the rows in a column, containing NaN value.

Also

    .notnull()

does the opposite. It finds the rows not containing NaN values.

The command 

    .isnull().sum()

counts the number of NaN values in a dataset.

The command 

    .fillna(0)
    .fillna('Unknown')
    .fillna(df['specific_column'].mean())

fills the empty data with either 0, or 'Unknown' or the mean value of that column.

If we want to fill a NaN data based on the previous data in that column, we use 

    .ffill() # Forward fill
and to do the opposite, we use

    .bfill() # Backward fill

* Whenever we have a few NaN elements and we are deciding to drop them we can use:

```python
coffee.dropna() # drops all rows with any missing values (NaN) in the coffee dataframe and returns a new dataframe without modifying the original dataframe
```

* which drops the **whole row** containing NaN. Yet if we want to just drop the whole row which contains NaN in a specific column we do:

```python
coffee.dropna(subset= ['Day']) # drops all rows with missing values (NaN) in the 'Day' column of the coffee dataframe and returns a new dataframe without modifying the original dataframe
```

* By default .dropna() removes the whole row, but if we decide to drop the column:

```python
coffee.dropna(axis=1) # drops any columns that contain missing values (NaN) from the coffee dataframe  
```
